```python
import pandas as pd

from scouting.data_transform.xthreat import compute_xthreat

spadl_data = pd.read_parquet("../tests/spadl_salzbourg_sb29_game_data.parquet")
spadl_data = compute_xthreat(spadl_data)
spadl_data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(spadl_data["xT_value"])
plt.title("Distribution of xT Values")
plt.xlabel("xT Value")
plt.ylabel("Density")
plt.show()
```

```python
high_xthreat_df = spadl_data[spadl_data["xT_value"].abs()>0.1]
sns.displot(high_xthreat_df["xT_value"])
plt.title("Distribution of xT Values")
plt.xlabel("xT Value")
plt.ylabel("Density")
plt.show()
```

```python
interesting_sequences = high_xthreat_df['sequence_id'].unique().tolist()
interesting_sequences
```

```python
import matplotsoccer
from scouting.visualisations.misc import plot_events_on_pitch


for sequence_id in interesting_sequences[:15]:
    print(f"\n=================================\n\nSequence {sequence_id}:\n")
    sequence_df = spadl_data[spadl_data["sequence_id"] == sequence_id].copy()

    plot_events_on_pitch(sequence_df, 'uefa')

    # From socceraction public notebooks
    # Print the game date and timestamp of the goal
    first_row = sequence_df.iloc[0].copy()
    game_info = f"{first_row.date} {first_row.home_team} {first_row.home_score}-{first_row.away_score} {first_row.away_team} {first_row.minute + 1}'"
    print(game_info)

    # Plot the actions
    def nice_time(row):
        minute = row.minute
        return f"{minute}'"

    sequence_df["nice_time"] = sequence_df.apply(nice_time, axis=1)

    labels = sequence_df[["nice_time", "type_name", "player", "team", "xT_value"]]
    
    ax = matplotsoccer.actions(
        location=sequence_df[["start_x", "start_y", "end_x", "end_y"]],
        action_type=sequence_df["type_name"],
        team= sequence_df["team"],
        result= sequence_df["result_name"] == "success",
        label=labels,
        labeltitle=["time", "actiontype", "player", "team", "xThreat"],
        zoom=False,
        figsize=6
    )
```
